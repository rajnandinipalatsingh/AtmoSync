```python
import kagglehub
import shutil
import os


path = kagglehub.dataset_download(
    "ankushnarwade/indian-climate-dataset-20242025"
)

raw_path = r"C:\PlayGround\Infotact Data Analytics\AtmoSync\dataset\raw"

os.makedirs(raw_path, exist_ok=True)

for file_name in os.listdir(path):
    source = os.path.join(path, file_name)
    destination = os.path.join(raw_path, file_name)

    if os.path.isfile(source):
        shutil.copy2(source, destination)

print("Dataset saved to:", os.path.abspath(raw_path))
```
