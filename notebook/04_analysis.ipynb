### Objective

Convert exploratory findings into measurable insights that can be used for environmental interpretation and dashboard reporting.

```python
# ============================================================
# AtmoSync - 04: Environmental Analysis and Insights
# ============================================================

# Import required libraries
import pandas as pd
import numpy as np

# Import Path for reliable file paths
from pathlib import Path


# ============================================================
# 1. DEFINE PROJECT PATH
# ============================================================

# Get the project root directory
# This assumes the notebook is inside the "notebooks" folder.
PROJECT_ROOT = Path.cwd().parent


# ============================================================
# 2. DEFINE PROCESSED DATASET PATH
# ============================================================

PROCESSED_DATA_PATH = (
    PROJECT_ROOT
    / "dataset"
    / "processed"
    / "microclimate_cleaned.csv"
)


# ============================================================
# 3. LOAD PROCESSED DATASET
# ============================================================

# Load the cleaned dataset
df = pd.read_csv(
    PROCESSED_DATA_PATH,
    parse_dates=["cdTimestamp"]
)


# ============================================================
# 4. VERIFY DATA
# ============================================================

print("Dataset loaded successfully.")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nFirst 5 records:")
display(df.head())
```

```python
# ============================================================
# 5. CHECK DATA TYPES
# ============================================================

df.info()
```

```python
# ============================================================
# 6. CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

print("Missing values:")
display(missing_values[missing_values > 0])
```

```python
# ============================================================
# 7. BASIC STATISTICS
# ============================================================

df.describe()
```

```python
lulc_summary = (
    df.groupby("LULC_Type")
    .agg(
        Average_Temperature=(
            "Temperature",
            "mean"
        ),

        Average_Humidity=(
            "Humidity",
            "mean"
        ),

        Average_Wind_Speed=(
            "Wind_Speed",
            "mean"
        ),

        Average_Solar_Radiation=(
            "Solar_Radiation",
            "mean"
        ),

        Average_NDVI=(
            "NDVI",
            "mean"
        )
    )
    .round(2)
)

lulc_summary
```

```python
location_summary = (
    df.groupby("Location_ID")
    .agg(
        Average_Temperature=(
            "Temperature",
            "mean"
        ),

        Maximum_Temperature=(
            "Temperature",
            "max"
        ),

        Average_Humidity=(
            "Humidity",
            "mean"
        ),

        Average_NDVI=(
            "NDVI",
            "mean"
        )
    )
    .round(2)
)

location_summary.sort_values(
    "Average_Temperature",
    ascending=False
).head(10)
```

```python
hottest = df.loc[
    df["Temperature"].idxmax()
]

hottest
```

```python
highest_ndvi = df.loc[
    df["NDVI"].idxmax()
]

highest_ndvi
```

```python
# ============================================================
# Number of locations and LULC types
# ============================================================

print("Number of Locations:", df["Location_ID"].nunique())
print("Number of LULC Types:", df["LULC_Type"].nunique())
```

```python
# ============================================================
# Overall KPI values
# ============================================================

kpis = {
    "Average Temperature": df["Temperature"].mean(),
    "Maximum Temperature": df["Temperature"].max(),
    "Minimum Temperature": df["Temperature"].min(),
    "Average Humidity": df["Humidity"].mean(),
    "Average Wind Speed": df["Wind_Speed"].mean(),
    "Average Solar Radiation": df["Solar_Radiation"].mean(),
    "Average NDVI": df["NDVI"].mean(),
}

for name, value in kpis.items():
    print(f"{name}: {value:.2f}")
```

```python
# ============================================================
# TEMPERATURE DISTRIBUTION
# ============================================================

temperature_stats = df["Temperature"].describe()

display(temperature_stats)
```

```python
# ============================================================
# TEMPERATURE RANGE
# ============================================================

temperature_range = (
    df["Temperature"].max()
    - df["Temperature"].min()
)

print(f"Temperature Range: {temperature_range:.2f} °C")
```

```python
# ============================================================
# EXTREME TEMPERATURE OBSERVATIONS
# ============================================================

extreme_temperature = df[
    (df["Temperature"] == df["Temperature"].min()) |
    (df["Temperature"] == df["Temperature"].max())
]

display(
    extreme_temperature[
        [
            "cdTimestamp",
            "Location_ID",
            "LULC_Type",
            "Temperature",
            "Humidity",
            "Solar_Radiation",
            "NDVI"
        ]
    ]
)
```

```python
# ============================================================
# LOW TEMPERATURE OBSERVATIONS
# ============================================================

low_temperature = (
    df[df["Temperature"] <= -5]
    .sort_values("Temperature")
)

print(
    f"Observations with Temperature <= -5°C: "
    f"{len(low_temperature)}"
)

display(
    low_temperature[
        [
            "cdTimestamp",
            "Location_ID",
            "LULC_Type",
            "Temperature",
            "Humidity",
            "Solar_Radiation",
            "NDVI"
        ]
    ].head(20)
)
```

```python
# ============================================================
# LOW-TEMPERATURE OBSERVATIONS BY LOCATION
# ============================================================

low_temp_by_location = (
    df[df["Temperature"] <= -5]
    .groupby("Location_ID")
    .agg(
        Low_Temperature_Count=("Temperature", "count"),
        Average_Low_Temperature=("Temperature", "mean"),
        Minimum_Temperature=("Temperature", "min")
    )
    .sort_values(
        "Low_Temperature_Count",
        ascending=False
    )
    .round(2)
)

display(low_temp_by_location)
```

```python
# ============================================================
# LOW-TEMPERATURE OBSERVATIONS BY LULC
# ============================================================

low_temp_by_lulc = (
    df[df["Temperature"] <= -5]
    .groupby("LULC_Type")
    .agg(
        Low_Temperature_Count=("Temperature", "count"),
        Average_Low_Temperature=("Temperature", "mean"),
        Minimum_Temperature=("Temperature", "min")
    )
    .sort_values(
        "Low_Temperature_Count",
        ascending=False
    )
    .round(2)
)

display(low_temp_by_lulc)
```

```python
OUTPUT_PATH = (
    PROJECT_ROOT
    / "dataset"
    / "processed"
    / "lulc_summary.csv"
)

lulc_summary.to_csv(
    OUTPUT_PATH
)

print("LULC summary exported.")
```
