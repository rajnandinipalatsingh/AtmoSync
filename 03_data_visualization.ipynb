```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import folium
```

```python
df = pd.read_csv("../dataset/processed/cleanedDataset.csv")
```

```python
plt.plot(df["date"], df['NDVI'])
plt.xlabel('Date')
plt.ylabel('NDVI')
plt.title('NDVI Over Time')
plt.show()
```

```python
sns.heatmap(
    df[['Temperature', 'Humidity', 'Precipitation',
        'Solar_Radiation', 'Air_Pressure', 'NDVI']].corr(),
    annot=True
)
plt.show()
```

```python
sns.scatterplot(data=df, x='Temperature', y='NDVI')
```

```python
fig = px.scatter(
    df,
    x='Temperature',
    y='NDVI',
    color='LULC_Type',
    title='Temperature vs NDVI'
)

fig.show()
```
